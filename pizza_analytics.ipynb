## Group by analysis

```python
#Total  quantity sold in each pizza category
# in which base/condtion  we are categoryiing that write in () this bracket 
a=df.groupby('pizza_category')['quantity'].sum()
print(a)
```

```python
#Total quantity sold and total  revenue in each pizza category

a=df.groupby('pizza_category')[['quantity','total_price']].sum()
print(a)
```

```python
#Total orders in each quantity (Based on order id)
#In your data  same order id is repeated  (ex: 2,0,2,0,2.0...  then use nunique etc)
#correct method is using nunique
k=df.groupby('quantity')['order_id'].nunique()
print(k)
```

```python
#If we use count it comes like this 
k=df.groupby('quantity')['order_id'].count()
print(k)
```

```python
#calculate total revenue for every month
k=df.groupby('month_name')['total_price'].sum()
print(k)
```

```python
#calculate total orders for every day of week
k=df.groupby('day_name')['order_id'].sum()
print(k)
```

```python
#calculate  total quantity sold in every hour
qty=df.groupby('hour')['quantity'].sum()
print(qty)
#plotting the graph
m=qty.index
n=qty.values
#plt.plot(m,n)
# if you want markers you can use that also 
plt.plot(m,n,'rD--')
plt.title("Total Quantity sold in every hours")
plt.xlabel("Hour")
plt.ylabel("Total quantity sold")
plt.show()
```

```python
#Top 5 pizzas based on a total revenue
#Method 1:

# to sort use sort_values()
top=df.groupby('pizza_name')['total_price'].sum().sort_values().tail(5)
print(top)
```

```python
#Method 2:

top=df.groupby('pizza_name')['total_price'].sum().sort_values(ascending=False).head(5)
print(top)
```

```python
#Do the grahp for above question that is( pizzas based on a total revenue)
x=top.index
y=top.values
plt.barh(x,y, color='purple')
plt.title("Top 5 pizzas based on total revenue")
plt.xlabel("Total revenue")
plt.ylabel("Pizza name")
plt.show()
```
